## Exercise 3: The code below does the following for $n=3$. Do the same for $n=4,5$.
- ### Check that the row orthogonality theorem holds.

```python
irr = S3.irreducible_characters()
#irr is the list of irreducible characters. Since it is a list, we can access each of these three elements as irr[0], irr[1], irr[2].
display(irr[0].values())
display(irr[1].values())
display(irr[2].values())
```

```python
#Now use .scalar_product to perform the inner product of one character with another.
display(irr[0].scalar_product(irr[0]))
display(irr[0].scalar_product(irr[1]))
display(irr[0].scalar_product(irr[2]))

display(irr[1].scalar_product(irr[0]))
display(irr[1].scalar_product(irr[1]))
display(irr[1].scalar_product(irr[2]))

display(irr[2].scalar_product(irr[0]))
display(irr[2].scalar_product(irr[1]))
display(irr[2].scalar_product(irr[2]))
#Need to check the scalar product for all combinations of irr[i], irr[j] for values of i,j between 0 and 2. You'll want to write a loop for higher n, since there will be many more combinations for S_4, S_5, and S_6.
```

For $S_{4},S_{5}$ I show it in matrix form where $M=(\langle\chi_{i},\chi_{j}\rangle)=(\delta(i,j))=I$

$$S_{4}$$

```python
irr = S4.irreducible_characters()

for i in range(len(irr)):
    display(irr[i].values())

print("------")

M = matrix([
    [irr[i].scalar_product(irr[j]) for j in range(len(irr))]
    for i in range(len(irr))
])

display(M)
```

$$S_{5}$$

```python
irr = S5.irreducible_characters()

for i in range(len(irr)):
    display(irr[i].values())

print("------")

M = matrix([
    [irr[i].scalar_product(irr[j]) for j in range(len(irr))]
    for i in range(len(irr))
])

display(M)
```

The below cells show that we get $|S_{4}|=24=4!,|S_{5}|=5!=120$, respectively, as expected, when summing the squares of the dimensions of each irreducible representation.

```python
S4 = SymmetricGroup(4)
irr = S4.irreducible_characters()

sum(list(irr[i])[0]^2 for i in range(len(irr)))
```
