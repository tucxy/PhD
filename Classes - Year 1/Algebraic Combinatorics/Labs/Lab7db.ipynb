p.196 Permutation Group Generators:

```python
e=Permutation([(1),(2),(3),(4),(5),(6)])
g1 = Permutation([(2, 3, 4, 5)])
g2 = Permutation([(2, 4), (3, 5)])
g3 = Permutation([(1, 4), (2, 6), (3, 5)])
g4= Permutation([(1,2,5), (3,6,4)])
G = PermutationGroup([e,g1,g2,g3,g4])
display(G)
```

```python
G.conjugacy_classes()
```

```python
dplay(G.cycle_index())
dplay(G.profile_series())
```

Plugging in 1 to the profile series will give you the number of orbits of all possible n-element subsets in this action, which which is just the total number of orbits since the set being acted on here is finite. The result is confirmed on the next page.

```python

dplay(G.profile_series().subs(z=1))
```
