## Exercise 4: Finish the 'for' loop below to check that for every involution of $S_3$, its corresponding insertion and recording tableaux are equal. Do the same for $n=4,5,6$.

```python
invcount = 0
eqcount = 0

for i in [3,4,5,6]:
    for sigma in Permutations(i):
        if sigma == sigma.inverse():
            invcount += 1
            P, Q = RSK(sigma)
            eqcount += int(P==Q)
    dplay(f"It is {invcount==eqcount} that for every involution of S_{i}, its corresponding insertion and recording tableaux are equal")
```

You can see the loop above checks that statement given for $S_{3},\dots,S_{6}$ by counting the number of inversions and inversions where the statement holds and then comparing the two at the end.

## Exercise 5: Calculate the sum of $(f^{\lambda})^2$ for all $\lambda$ with $3$ boxes. Do the same for $n=4,5,6$. Conjecture a formula in terms of $n$. How does RSK prove this formula? (The answer is in Section 4.3.)

```python
for i in [3,4,5,6]:
    totsum = 0

    for labda in Partitions(i):
        f = StandardTableaux(labda).cardinality()
        totsum = totsum + f^2

    print(f"It is {totsum == factorial(i)} that {i}! = Sum of (f^lambda)^2 for all lambda with {i} boxes.")
```

RSK proves this because it gives a correspondence between permutations $\sigma\in S_{n}$ to pairs $(P,Q)$ of standard young tableau of the same shape $\lambda\vdash n$. (I am assuming we proved this but if not, we saw that there exists a bijective inverse for the mapping earlier in the lab.) So of course the number $(f^{\lambda})^{2}$ of such pairs is equal to the number of such permutations $\sigma\in S_{n}$. That is:
$$(f^{\lambda})^{2}= |S_{n}|=n!$$
